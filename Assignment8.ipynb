```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv(r"C:\Users\dell\Documents\ML\Spending_Score.csv")
```

```python
df
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
X = df[['Age','Annual_Income','Spending_Score']]
```

```python
wcss = []

for i in range(1,11):
    model = KMeans(n_clusters=i, random_state=42)
    model.fit(X)
    wcss.append(model.inertia_)

plt.figure(figsize=(7,5))
plt.plot(range(1,11), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.grid(True)
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=42)

df['Cluster'] = kmeans.fit_predict(X)
```

```python
print("Customer Segments")
print(df['Cluster'])

print("Cluster Centers")
print(kmeans.cluster_centers_)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df['Annual_Income'],
    df['Spending_Score'],
    c=df['Cluster'],
    cmap='rainbow',
    s=70
)

plt.scatter(
    kmeans.cluster_centers_[:,1],
    kmeans.cluster_centers_[:,2],
    color='black',
    marker='X',
    s=250,
    label='Centroids'
)
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Segmentation")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import  numpy as np
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from sklearn.cluster import KMeans
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
df
```

```python
df.shape
```

```python
X = df[['Income', 'Credit_Score', 'Loan_Amount']]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
hierarchical = AgglomerativeClustering(
    n_clusters=3,
    metric='euclidean',
    linkage='ward'
)

hierarchical_labels = hierarchical.fit_predict(X_scaled)

```

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42
)

kmeans_labels = kmeans.fit_predict(X_scaled)


df['KMeans_Cluster'] = kmeans_labels


print("K-Means Results:")
print(df.head())
```

```python

comparison = df[
    ['Income',
     'Credit_Score',
     'Loan_Amount',
     'Hierarchical_Cluster',
     'KMeans_Cluster']
]
print("Cluster Comparison:")
print(comparison.head(20))
```

```python
print("Hierarchical Cluster Distribution:")
print(df['Hierarchical_Cluster'].value_counts())


print("K-Means Cluster Distribution:")
print(df['KMeans_Cluster'].value_counts())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\dell\Documents\ML\Customer.csv")
```

```python
df
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
X = df[['Age', 'Annual_Income', 'Spending_Score']]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
wcss = []

for k in range(1, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))
plt.plot(range(1,11), wcss, marker='o')

plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method")

plt.grid()
plt.show()
```

```python
df['Cluster'] = kmeans.predict(X_scaled)


print("\nCustomer Cluster Labels:")
print(df[['Age','Annual_Income','Spending_Score','Cluster']])
```

```python
centers = scaler.inverse_transform(
    kmeans.cluster_centers_
)

cluster_centers = pd.DataFrame(
    centers,
    columns=['Age','Annual_Income','Spending_Score']
)

print("\nCluster Centers:")
print(cluster_centers)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df['Annual_Income'],
    df['Spending_Score'],
    c=df['Cluster'],
    cmap='rainbow',
    s=80
)
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Segmentation using K-Means")

plt.legend()
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(12,5))


# Agglomerative Plot

plt.subplot(1,2,1)

plt.scatter(
    df['Area_sqft'],
    df['Price'],
    c=df['Agglomerative_Cluster'],
    cmap='viridis',
    s=70
)
plt.title("Agglomerative Clustering")
plt.xlabel("Area (sqft)")
plt.ylabel("Price")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
df = pd.read_csv(r"C:\Users\dell\Documents\ML\customer_finance.csv")
```

```python
df
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
X = df[
    [
        'Age',
        'Income',
        'Spending_Score',
        'Savings',
        'Credit_Score'
    ]
]
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)


```

```python
pca_df = pd.DataFrame(
    X_pca,
    columns=[
        'Principal_Component_1',
        'Principal_Component_2'
    ]
)


print("\nReduced Dataset:")
print(pca_df.head())
```

```python
print("\nExplained Variance Ratio:")

print(pca.explained_variance_ratio_)


print("\nTotal Variance Captured:")

print(
    sum(pca.explained_variance_ratio_)
)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    pca_df['Principal_Component_1'],
    pca_df['Principal_Component_2'],
    c='blue',
    s=70
)


plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.title(
    "PCA Visualization of Customer Financial Data"
)
plt.grid()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv(r"C:\Users\dell\Documents\ML\students.csv")
```

```python
df
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
X = df[
    [
        'Study_Hours',
        'Attendance',
        'Internal_Marks',
        'Project_Score'
    ]
]
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)
kmeans.fit(X_scaled)
```

```python
df['Student_Group'] = kmeans.predict(X_scaled)
print("\nStudent Groups:")

print(
    df[
        [
            'Study_Hours',
            'Attendance',
            'Internal_Marks',
            'Project_Score',
            'Student_Group'
        ]
    ]
)

```

```python
centers = scaler.inverse_transform(
    kmeans.cluster_centers_
)


cluster_centers = pd.DataFrame(
    centers,
    columns=[
        'Study_Hours',
        'Attendance',
        'Internal_Marks',
        'Project_Score'
    ]
)


print("\nCluster Centers:")

print(cluster_centers)
```

```python
plt.figure(figsize=(8,6))


plt.scatter(
    df['Study_Hours'],
    df['Internal_Marks'],
    c=df['Student_Group'],
    cmap='viridis',
    s=80
)
plt.scatter(
    cluster_centers['Study_Hours'],
    cluster_centers['Internal_Marks'],
    color='red',
    marker='X',
    s=250,
    label='Cluster Centers'
)


plt.xlabel("Study Hours")
plt.ylabel("Internal Marks")

plt.title(
    "University Student Clustering using K-Means"
)

plt.legend()
plt.grid()

plt.show()
```
