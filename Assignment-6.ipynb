```python
import pandas as pd
data = {
    "Department": ["IT", "HR", "IT", "Sales", "HR"],
    "Salary": [50000, 30000, 60000, 40000, 35000]
}
df = pd.DataFrame(data)
avg_salary = df.groupby("Department")["Salary"].mean()
print(avg_salary)
```

```python
#1. Find the average salary for each department
import pandas as pd
avg_salary = df.groupby("Department")["Salary"].mean()
print(avg_salary)
```

```python
#2. Find the maximum salary in each department
max_salary = df.groupby("Department")["Salary"].max()
print(max_salary)
```

```python
#3. Find total salary paid by each department
total_salary = df.groupby("Department")["Salary"].sum()
print(total_salary)
```

```python
#4. Count the number of employees in each department
employee_count = df.groupby("Department").size()
print(employee_count)
```

```python
#5.	Find the average salary and average experience for each department. 
Merge & Join 
df = pd.DataFrame(data)
result = df.groupby("Department")[["Salary", "Experience"]].mean()
print(result)
```

```python
#6.	Perform an inner join between two DataFrames on a common column.
import pandas as pd
df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ram", "Shyam", "Mohan"]
})
df2 = pd.DataFrame({
    "ID": [1, 2, 4],
    "City": ["Pune", "Mumbai", "Delhi"]
})
result = pd.merge(df1, df2, on="ID", how="inner")
print(result)
```

```python
#7.	Perform a left join between two DataFrames
result = pd.merge(df1, df2, on="ID", how="left")
print(result)
```

```python
#8.	Perform a right join between two DataFrames. 
result = pd.merge(df1, df2, on="ID", how="right")
print(result)
```

```python
#9.	Perform a full outer join between two DataFrames. 
result = pd.merge(df1, df2, on="ID", how="outer")
print(result)
```

```python
#10.Find records that exist in one DataFrame but not in the other. 
#Concatenation
result = pd.merge(df1, df2, on="ID", how="outer", indicator=True)
print(result[result["_merge"] != "both"])
```

```python
#11.Concatenate two DataFrames row-wise. 
result = pd.concat([df1, df1], axis=0)
print(result)
```

```python
#12.Concatenate two DataFrames column-wise. 
import pandas as pd
df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ram", "Shyam", "Mohan"]
})
df3 = pd.DataFrame({
    "Salary": [50000, 60000, 55000]
})
result = pd.concat([df1, df3], axis=1)
print(result)
```

```python
#13.Concatenate multiple DataFrames and reset the index. 
result = pd.concat([df1, df1], ignore_index=True)
print(result)
```

```python
#14.	Concatenate DataFrames using keys to create a hierarchical index. 
#Pivot Tables
result = pd.concat([df1, df2], keys=["First", "Second"])
print(result)
```

```python
import pandas as pd
data = {
    "Region": ["East", "West", "East", "West", "North"],
    "Product": ["Laptop", "Mobile", "Laptop", "Mobile", "Laptop"],
    "Sales": [50000, 30000, 40000, 25000, 60000],
    "Date": ["2026-01-01", "2026-01-02", "2026-01-03",
             "2026-01-04", "2026-01-05"]
}
df = pd.DataFrame(data)
```

```python
#15.Create a pivot table showing total sales by region.
pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    aggfunc="sum"
)
print(pivot)
```

```python
#16.Create a pivot table with regions as rows and products as columns
pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    columns="Product",
    aggfunc="sum"
)
print(pivot)
```

```python
#17.	Create a pivot table showing average sales for each product in each region. 
#Time Series Basics
pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    columns="Product",
    aggfunc="mean"
)
print(pivot)
```

```python
#21.Find the total sales for each product using GroupBy. 
import pandas as pd
sales = {
    'Product': ['Laptop', 'Mobile', 'Laptop', 'Tablet', 'Mobile'],
    'Sales': [50000, 20000, 30000, 15000, 25000]
}
df = pd.DataFrame(sales)
result = df.groupby('Product')['Sales'].sum()
print(result)
```

```python
#22.Find the highest-selling product in each region
import pandas as pd
data = {
    'Region': ['North', 'North', 'South', 'South'],
    'Product': ['Laptop', 'Mobile', 'Laptop', 'Mobile'],
    'Sales': [50000, 70000, 60000, 55000]
}
df = pd.DataFrame(data)
result = df.loc[df.groupby('Region')['Sales'].idxmax()]
print(result)
```

```python
#25.Create a pivot table showing monthly sales by product. 
import pandas as pd
data = {
    'Month': ['Jan', 'Jan', 'Feb', 'Feb'],
    'Product': ['Laptop', 'Mobile', 'Laptop', 'Mobile'],
    'Sales': [50000, 30000, 60000, 35000]
}
df = pd.DataFrame(data)
pivot = pd.pivot_table(
    df,
    values='Sales',
    index='Month',
    columns='Product',
    aggfunc='sum'
)
print(pivot)
```

```python
#26.Resample daily sales data into weekly sales totals. 
import pandas as pd
data = {
    'Date': pd.date_range('2025-01-01', periods=10),
    'Sales': [100, 200, 150, 300, 250, 400, 350, 450, 500, 550]
}
df = pd.DataFrame(data)
df.set_index('Date', inplace=True)
weekly_sales = df.resample('W').sum()
print(weekly_sales)
```

```python
#27.Calculate cumulative sales over time. 
import pandas as pd
data = {
    'Sales': [1000, 2000, 1500, 3000, 2500]
}
df = pd.DataFrame(data)
df['CumulativeSales'] = df['Sales'].cumsum()
print(df)
```

```python
#28.Find the top 3 departments based on average salary. 
import pandas as pd
data = {
    'Department': ['IT', 'HR', 'Finance', 'IT', 'Finance', 'Sales'],
    'Salary': [50000, 30000, 60000, 55000, 65000, 40000]
}
df = pd.DataFrame(data)
result = (
    df.groupby('Department')['Salary']
    .mean()
    .sort_values(ascending=False)
    .head(3)
)
print(result)
```

```python
#29.Merge three DataFrames using a common key. 
import pandas as pd
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Amit', 'Priya', 'Rahul']
})
df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'City': ['Pune', 'Mumbai', 'Nagpur']
})
df3 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [85, 90, 78]
})
result = pd.merge(df1, df2, on='ID')
result = pd.merge(result, df3, on='ID')
print(result)
```

```python
#30.Create a pivot table with multiple aggregation functions (sum, mean, count).
import pandas as pd
data = {
    'Department': ['IT', 'IT', 'HR', 'HR', 'Finance'],
    'Salary': [50000, 60000, 40000, 45000, 70000]
}
df = pd.DataFrame(data)
pivot = pd.pivot_table(
    df,
    values='Salary',
    index='Department',
    aggfunc=['sum', 'mean', 'count']
)
print(pivot)
```
