```python
#25. Create DataFrame with missing values and display missing values
import pandas as pd
import numpy as np
df = pd.DataFrame({
    'Name': ['Rahul', 'Priya', None],
    'Age': [22, np.nan, 23],
    'Marks': [78, 85, np.nan]
})
print(df.isnull())
```

```python
#26. Count missing values in each column
print(df.isnull().sum())
```

```python
#27. Check whether each column contains any missing values
print(df.isnull().any())
```

```python
#28. Fill missing Age with average age
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
#29. Fill missing Marks with 0
df['Marks'] = df['Marks'].fillna(0)
```

```python
#30. Remove all rows containing missing values
df.dropna(inplace=True)
print(df)
```

```python
#31. Create the Product DataFrame
import pandas as pd
data = {
    'Product': ['Laptop', 'Mobile', 'Chair', 'Table', 'Headphones'],
    'Category': ['Electronics', 'Electronics', 'Furniture', 'Furniture', 'Electronics'],
    'Price': [50000, 20000, 3000, 5000, 2000],
    'Quantity': [5, 10, 8, 4, 15]
}
df = pd.DataFrame(data)
print(df)
```

```python
#32. Display only Electronics products
print(df[df['Category'] == 'Electronics'])
```

```python
#33. Display products whose price is greater than 10000
print(df[df['Price'] > 10000])
```

```python
#34. Find the average price
print(df['Price'].mean())
```

```python
#35. Find the most expensive product
print(df.loc[df['Price'].idxmax()])
```

```python
#36. Find the least expensive product
print(df.loc[df['Price'].idxmin()])
```

```python
#37. Display products whose quantity is greater than 7
print(df[df['Quantity'] > 7])
```

```python
#38. Add Total_Price column
df['Total_Price'] = df['Price'] * df['Quantity']
print(df)
```

```python
#39. Display Product and Total_Price columns
print(df[['Product', 'Total_Price']])
```

```python
#40. Find total quantity of all products
print(df['Quantity'].sum())
```
